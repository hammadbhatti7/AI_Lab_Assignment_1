```python
# Student Result Management System
students = []

def calculate_result(marks):
    total = sum(marks)
    percentage = (total / 300) * 100

    if percentage >= 80:
        grade = "A"
    elif percentage >= 70:
        grade = "B"
    elif percentage >= 60:
        grade = "C"
    elif percentage >= 50:
        grade = "D"
    else:
        grade = "F"

    return total, percentage, grade


# Enter data of 5 students
for i in range(5):
    print("\nEnter details of student", i + 1)

    name = input("Enter student name: ")
    roll_number = input("Enter roll number: ")

    python_marks = int(input("Enter Python marks: "))
    ai_marks = int(input("Enter Artificial Intelligence marks: "))
    math_marks = int(input("Enter Mathematics marks: "))

    student = {
        "name": name,
        "roll_number": roll_number,
        "marks": [python_marks, ai_marks, math_marks]
    }

    students.append(student)


# Display results and save them in a file
highest_student = None
highest_percentage = -1

with open("results.txt", "w") as file:

    file.write("STUDENT RESULT MANAGEMENT SYSTEM\n")
    file.write("=" * 40 + "\n")

    for student in students:
        total, percentage, grade = calculate_result(student["marks"])

        result = (
            "\nName: " + student["name"] +
            "\nRoll Number: " + student["roll_number"] +
            "\nPython Marks: " + str(student["marks"][0]) +
            "\nArtificial Intelligence Marks: " + str(student["marks"][1]) +
            "\nMathematics Marks: " + str(student["marks"][2]) +
            "\nTotal Marks: " + str(total) + "/300" +
            "\nPercentage: " + str(round(percentage, 2)) + "%" +
            "\nGrade: " + grade + "\n"
        )

        print(result)
        file.write(result)
        file.write("-" * 40 + "\n")

    
        if percentage > highest_percentage:
            highest_percentage = percentage
            highest_student = student

    # Display the highest scorer
    print("\nHighest Scoring Student")
    print("Name:", highest_student["name"])
    print("Roll Number:", highest_student["roll_number"])
    print("Percentage:", round(highest_percentage, 2), "%")

    # Save highest scorer in the file
    file.write("\nHIGHEST SCORING STUDENT\n")
    file.write("Name: " + highest_student["name"] + "\n")
    file.write("Roll Number: " + highest_student["roll_number"] + "\n")
    file.write("Percentage: " + str(round(highest_percentage, 2)) + "%\n")

print("\nAll results have been saved in results.txt")

```

```python
def analyze_sentence(sentence):

    words = sentence.split()

    # Count total words
    total_words = len(words)

    # Count total characters
    total_characters = len(sentence)

    # Find longest and shortest words
    longest_word = max(words, key=len)
    shortest_word = min(words, key=len)

    # Count vowels
    vowels = "aeiouAEIOU"
    vowel_count = 0

    for char in sentence:
        if char in vowels:
            vowel_count += 1

    # Create a set of unique words
    unique_words = set()

    for word in words:
        unique_words.add(word.lower())

    # Sort words in alphabetical order
    alphabetical_words = sorted(unique_words)
    
    print("\nWord Analysis")
    print("Total Words:", total_words)
    print("Total Characters:", total_characters)
    print("Longest Word:", longest_word)
    print("Shortest Word:", shortest_word)
    print("Total Vowels:", vowel_count)
    print("Unique Words:", unique_words)
    print("Alphabetical Order:", alphabetical_words)

    # Save the analysis in a file
    with open("word_analysis.txt", "w") as file:
        file.write("WORD ANALYSIS REPORT\n")
        file.write("--------------------\n")
        file.write("Sentence: " + sentence + "\n")
        file.write("Total Words: " + str(total_words) + "\n")
        file.write("Total Characters: " + str(total_characters) + "\n")
        file.write("Longest Word: " + longest_word + "\n")
        file.write("Shortest Word: " + shortest_word + "\n")
        file.write("Total Vowels: " + str(vowel_count) + "\n")
        file.write("Unique Words: " + str(unique_words) + "\n")
        file.write("Alphabetical Order: " + str(alphabetical_words) + "\n")

    print("\nAnalysis saved in word_analysis.txt")



sentence = input("Enter a sentence: ")

if len(sentence.split()) > 0:
    analyze_sentence(sentence)
else:
    print("Please enter a valid sentence.")

```

```python
# Simple Shopping Program

products = []

def calculate_bill(products):
    total_bill = 0

    for product in products:
        product_total = product["price"] * product["quantity"]
        total_bill = total_bill + product_total

    discount = 0
    if total_bill > 10000:
        discount = total_bill * 0.10

    final_bill = total_bill - discount

    return total_bill, discount, final_bill


for i in range(5):
    print("\nEnter details of product", i + 1)

    name = input("Enter product name: ")
    price = float(input("Enter product price: Rs. "))
    quantity = int(input("Enter quantity: "))

    product = {
        "name": name,
        "price": price,
        "quantity": quantity
    }

    products.append(product)

total_bill, discount, final_bill = calculate_bill(products)


most_expensive = products[0]

for product in products:
    if product["price"] > most_expensive["price"]:
        most_expensive = product

# Display and save the bill
with open("bill.txt", "w") as file:

    print("\n========== SHOPPING BILL ==========")
    file.write("========== SHOPPING BILL ==========\n")

    for product in products:
        product_total = product["price"] * product["quantity"]

        result = (
            "\nProduct Name: " + product["name"] +
            "\nPrice: Rs. " + str(product["price"]) +
            "\nQuantity: " + str(product["quantity"]) +
            "\nTotal Price: Rs. " + str(product_total) + "\n"
        )

        print(result)
        file.write(result)

    print("Total Bill: Rs.", total_bill)
    print("Discount: Rs.", discount)
    print("Final Bill: Rs.", final_bill)

    file.write("\nTotal Bill: Rs. " + str(total_bill) + "\n")
    file.write("Discount: Rs. " + str(discount) + "\n")
    file.write("Final Bill: Rs. " + str(final_bill) + "\n")

    print("\nMost Expensive Product:", most_expensive["name"])
    print("Product Price: Rs.", most_expensive["price"])

    file.write("\nMost Expensive Product: " + most_expensive["name"] + "\n")
    file.write("Product Price: Rs. " + str(most_expensive["price"]) + "\n")

print("\nBill saved successfully in bill.txt")

```

```python
# Simple Calculator

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    return a / b

def modulus(a, b):
    return a % b

while True:
    num1 = float(input("\nEnter first number: "))
    num2 = float(input("Enter second number: "))

    print("\nSelect an operation:")
    print("+ for Addition")
    print("- for Subtraction")
    print("* for Multiplication")
    print("/ for Division")
    print("% for Modulus")

    operation = input("Enter operation: ")

    if operation == "+":
        result = add(num1, num2)
        print("Result:", result)

    elif operation == "-":
        result = subtract(num1, num2)
        print("Result:", result)

    elif operation == "*":
        result = multiply(num1, num2)
        print("Result:", result)

    elif operation == "/":
        if num2 != 0:
            result = divide(num1, num2)
            print("Result:", result)
        else:
            print("Error! Cannot divide by zero.")

    elif operation == "%":
        if num2 != 0:
            result = modulus(num1, num2)
            print("Result:", result)
        else:
            print("Error! Cannot calculate modulus by zero.")

    else:
        print("Invalid operation!")

    choice = input("\nDo you want another calculation? (yes/no): ")

    if choice.lower() != "yes":
        print("Calculator closed. Thank you!")
        break

```

```python
# Simple Restaurant Ordering System

# Restaurant menu
menu = {
    1: {"name": "Burger", "price": 500},
    2: {"name": "Pizza", "price": 800},
    3: {"name": "Fries", "price": 250},
    4: {"name": "Drink", "price": 150}
}
orders = []

def calculate_bill(orders):
    total_bill = 0

    for order in orders:
        total = order["price"] * order["quantity"]
        total_bill = total_bill + total

    return total_bill

while True:
    print("\n========== RESTAURANT MENU ==========")
    print("1. Burger - Rs. 500")
    print("2. Pizza  - Rs. 800")
    print("3. Fries  - Rs. 250")
    print("4. Drink  - Rs. 150")
    print("0. Finish Order")

    choice = int(input("Select an item (0 to finish): "))

    if choice == 0:
        break

    elif choice in menu:
        quantity = int(input("Enter quantity: "))

        if quantity > 0:
            order = {
                "name": menu[choice]["name"],
                "price": menu[choice]["price"],
                "quantity": quantity
            }

            orders.append(order)

            print(order["name"], "added to your order.")

        else:
            print("Quantity must be greater than zero.")

    else:
        print("Invalid choice! Please select from 1 to 4.")


# Calculate and display the final bill
if len(orders) > 0:
    total_bill = calculate_bill(orders)

    print("\n========== FINAL BILL ==========")

    with open("orders.txt", "w") as file:
        file.write("RESTAURANT FINAL BILL\n")
        file.write("---------------------\n")

        for order in orders:
            item_total = order["price"] * order["quantity"]

            print(order["name"], "-", order["quantity"],
                  "x Rs.", order["price"], "=", item_total)

            file.write(
                order["name"] + " - " +
                str(order["quantity"]) + " x Rs. " +
                str(order["price"]) + " = Rs. " +
                str(item_total) + "\n"
            )

        print("---------------------")
        print("Total Bill: Rs.", total_bill)

        file.write("---------------------\n")
        file.write("Total Bill: Rs. " + str(total_bill) + "\n")

    print("Bill saved successfully in orders.txt")

else:
    print("No items were ordered.")

```
